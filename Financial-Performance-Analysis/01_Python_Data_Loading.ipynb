```python
# Step 1: Import required libraries

import pandas as pd
import mysql.connector

print("Libraries imported successfully!")
```

```python
# Step 2: Connect to MySQL database

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Password",
    database="financial_performance"
)

print("MySQL connection successful!")
```

```python
# Step 3: Load Customers table
df_customers = pd.read_sql("SELECT * FROM CUSTOMERS",conn)
print("Customers loaded successfully")
print("Rows:", df_customers.shape[0])
print("Columns:",df_customers.shape[1])
```

```python
# Step 4: Load Vendors table

df_vendors = pd.read_sql(
    "SELECT * FROM vendors",
    conn
)

print("Vendors loaded successfully!")
print("Rows:", df_vendors.shape[0])
print("Columns:", df_vendors.shape[1])
```

```python
 # 5: Load Headcount table

df_headcount = pd.read_sql(
    "SELECT * FROM headcount",
    conn
)

print("Headcount loaded successfully!")
print("Rows:", df_headcount.shape[0])
print("Columns:", df_headcount.shape[1])
```

```python
# Step 6: Load Budget table

df_budget = pd.read_sql(
    "SELECT * FROM budget",
    conn
)

print("Budget loaded successfully!")
print("Rows:", df_budget.shape[0])
print("Columns:", df_budget.shape[1])
```

```python
# Step 7: Load Financial Transactions table

df_transactions = pd.read_sql(
    "SELECT * FROM financial_transactions",
    conn
)

print("Financial transactions loaded successfully!")
print("Rows:", df_transactions.shape[0])
print("Columns:", df_transactions.shape[1])
```

```python
# Step 8: Validate all loaded DataFrames

tables = {
    "customers": df_customers,
    "vendors": df_vendors,
    "headcount": df_headcount,
    "budget": df_budget,
    "financial_transactions": df_transactions
}

print("Python Data Loading Validation")
print("-" * 40)

for table_name, df in tables.items():
    print(
        f"{table_name}: "
        f"{df.shape[0]} rows × {df.shape[1]} columns"
    )
```

```python
# Step 9: Create DataFrame inventory

inventory = pd.DataFrame({
    "Table": [
        "customers",
        "vendors",
        "headcount",
        "budget",
        "financial_transactions"
    ],
    "DataFrame": [
        "df_customers",
        "df_vendors",
        "df_headcount",
        "df_budget",
        "df_transactions"
    ],
    "Rows": [
        df_customers.shape[0],
        df_vendors.shape[0],
        df_headcount.shape[0],
        df_budget.shape[0],
        df_transactions.shape[0]
    ],
    "Columns": [
        df_customers.shape[1],
        df_vendors.shape[1],
        df_headcount.shape[1],
        df_budget.shape[1],
        df_transactions.shape[1]
    ]
})

inventory
```

```python
# Step 10: Final Data Loading completeness check

expected_rows = {
    "customers": 400,
    "vendors": 120,
    "headcount": 200,
    "budget": 72,
    "financial_transactions": 10400
}

print("Final Data Loading Validation")
print("-" * 40)

all_passed = True

for table_name, expected in expected_rows.items():
    actual = tables[table_name].shape[0]

    if actual == expected:
        print(f"{table_name}: PASSED ({actual} rows)")
    else:
        print(
            f"{table_name}: FAILED "
            f"(Expected {expected}, Found {actual})"
        )
        all_passed = False

print("-" * 40)

if all_passed:
    print("DATA LOADING VALIDATION: PASSED")
else:
    print("DATA LOADING VALIDATION: FAILED")
```
