# 3. Outlier Analysis

Outlier analysis is performed to identify unusually large or small financial transactions that may materially influence statistical results.

The Interquartile Range (IQR) method is used because it is robust to non-normal distributions.

For each financial measure:

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1
- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

Observations outside these bounds are classified as potential outliers.

Outliers will be treated as observations requiring investigation rather than automatically removed.

```python
# Statistical Analysis — Cell 11
# IQR-based outlier analysis

outlier_results = []

for measure in financial_measures:

    series = df_stats[measure].dropna()

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outlier_count = (
        (series < lower_bound) |
        (series > upper_bound)
    ).sum()

    outlier_percentage = (
        outlier_count / len(series)
    ) * 100

    outlier_results.append({
        "measure": measure,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outlier_count,
        "outlier_percentage": outlier_percentage
    })

outlier_analysis = pd.DataFrame(outlier_results)

print("IQR-BASED OUTLIER ANALYSIS")
print("=" * 80)

display(outlier_analysis)
```

# 6. Business Unit Significance Testing

A one-way ANOVA will be used to test whether the mean transaction amount differs significantly across the three business units.

### Hypotheses

- **H₀ (Null Hypothesis):** The mean transaction amount is equal across all business units.
- **H₁ (Alternative Hypothesis):** At least one business unit has a different mean transaction amount.

### Significance Level

The analysis will use:

**α = 0.05**

If the p-value is below 0.05, the null hypothesis will be rejected.

Before interpreting the result, the ANOVA assumptions will also be considered.

```python
# Statistical Analysis — Cell 17
# One-way ANOVA: Transaction Amount by Business Unit

from scipy import stats

business_unit_groups = [
    group["amount"].dropna().values
    for _, group in df_stats.groupby("business_unit")
]

anova_statistic, anova_pvalue = stats.f_oneway(
    *business_unit_groups
)

alpha = 0.05

print("ONE-WAY ANOVA — TRANSACTION AMOUNT BY BUSINESS UNIT")
print("=" * 70)

print(f"F-statistic: {anova_statistic:.6f}")
print(f"p-value: {anova_pvalue:.6e}")
print(f"Significance level (α): {alpha}")

print("\nConclusion:")

if anova_pvalue < alpha:
    print(
        "Reject H₀ — statistically significant differences "
        "exist between at least one pair of business-unit means."
    )
else:
    print(
        "Fail to reject H₀ — no statistically significant "
        "difference in mean transaction amount was detected."
    )
```

# 10. Year-wise Significance Testing

An independent two-sample Welch's t-test is used to determine whether the mean transaction amount differs significantly between 2022 and 2023.

Welch's t-test is selected because it does not require the two groups to have equal population variances.

### Hypotheses

- **H₀ (Null Hypothesis):** The mean transaction amount is equal in 2022 and 2023.
- **H₁ (Alternative Hypothesis):** The mean transaction amount differs between 2022 and 2023.

### Significance Level

**α = 0.05**

A p-value below 0.05 indicates statistically significant evidence of a difference in mean transaction amount.

```python
# Statistical Analysis — Cell 25
# Welch's t-test: Transaction Amount between 2022 and 2023

amount_2022 = df_stats.loc[
    df_stats["year"] == 2022,
    "amount"
].dropna()

amount_2023 = df_stats.loc[
    df_stats["year"] == 2023,
    "amount"
].dropna()

t_statistic, p_value = stats.ttest_ind(
    amount_2022,
    amount_2023,
    equal_var=False
)

alpha = 0.05

print("WELCH'S T-TEST — 2022 VS 2023")
print("=" * 70)

print(f"2022 observations: {len(amount_2022)}")
print(f"2023 observations: {len(amount_2023)}")

print(f"\n2022 mean: {amount_2022.mean():.6f}")
print(f"2023 mean: {amount_2023.mean():.6f}")

print(f"\nt-statistic: {t_statistic:.6f}")
print(f"p-value: {p_value:.6e}")
print(f"Significance level (α): {alpha}")

print("\nConclusion:")

if p_value < alpha:
    print(
        "Reject H₀ — the mean transaction amount "
        "differs significantly between 2022 and 2023."
    )
else:
    print(
        "Fail to reject H₀ — no statistically significant "
        "difference in mean transaction amount was detected."
    )
```

# 14. Financial Flow Distribution Assessment

The distribution of transaction amounts across financial-flow groups will be assessed before selecting a significance test.

The three groups are:

- Income
- Cost
- Balance Sheet

The assessment will use the Shapiro-Wilk test to evaluate normality within each group.

### Hypotheses

- H₀: The observations are consistent with a normal distribution.
- H₁: The observations are not consistent with a normal distribution.

### Significance Level

α = 0.05

Because the dataset contains a large number of observations, statistical significance will be interpreted cautiously and alongside the descriptive statistics.

```python
# Statistical Analysis — Cell 33
# Normality assessment by financial flow

print("SHAPIRO-WILK NORMALITY TEST — FINANCIAL FLOW")
print("=" * 80)

normality_results = []

for flow, group in df_stats.groupby("financial_flow"):

    values = group["amount"].dropna()

    statistic, p_value = stats.shapiro(values)

    normality_results.append({
        "financial_flow": flow,
        "observations": len(values),
        "shapiro_statistic": statistic,
        "p_value": p_value
    })

normality_analysis = pd.DataFrame(normality_results)

display(normality_analysis)

print("\nInterpretation:")

for _, row in normality_analysis.iterrows():

    if row["p_value"] < 0.05:
        result = "Reject H₀ — evidence against normality"
    else:
        result = "Fail to reject H₀ — no significant evidence against normality"

    print(
        f'{row["financial_flow"]}: '
        f'p-value = {row["p_value"]:.6e} → {result}'
    )
```

# 15. Financial Flow Significance Testing

A Kruskal–Wallis H test will be used to determine whether the distribution of transaction amounts differs significantly across the three financial-flow groups.

The Kruskal–Wallis test is selected because the normality assessment indicates that the Cost and Income groups do not follow a normal distribution.

### Hypotheses

- **H₀ (Null Hypothesis):** The transaction amount distributions are the same across all financial-flow groups.
- **H₁ (Alternative Hypothesis):** At least one financial-flow group has a different distribution.

### Significance Level

**α = 0.05**

If the p-value is below 0.05, H₀ will be rejected.

A significant result indicates that at least one group differs, but it does not identify which specific groups differ.

```python
# Statistical Analysis — Cell 35
# Kruskal-Wallis test: Transaction Amount by Financial Flow

financial_flow_groups = [
    group["amount"].dropna().values
    for _, group in df_stats.groupby("financial_flow")
]

kruskal_statistic, kruskal_pvalue = stats.kruskal(
    *financial_flow_groups
)

alpha = 0.05

print("KRUSKAL-WALLIS TEST — TRANSACTION AMOUNT BY FINANCIAL FLOW")
print("=" * 80)

print(f"H-statistic: {kruskal_statistic:.6f}")
print(f"p-value: {kruskal_pvalue:.6e}")
print(f"Significance level (α): {alpha}")

print("\nConclusion:")

if kruskal_pvalue < alpha:
    print(
        "Reject H₀ — statistically significant differences "
        "exist among the financial-flow distributions."
    )
else:
    print(
        "Fail to reject H₀ — no statistically significant "
        "difference among the financial-flow distributions was detected."
    )
```

# 16. Financial Flow Pairwise Post-hoc Analysis

Because the Kruskal–Wallis test identified a statistically significant difference among the financial-flow groups, pairwise comparisons will be performed to determine which specific groups differ.

The Mann–Whitney U test will be used for pairwise comparisons because it is a non-parametric test suitable for independent groups.

The three pairwise comparisons are:

1. Income vs Cost
2. Income vs Balance Sheet
3. Cost vs Balance Sheet

A Bonferroni correction will be applied to control the family-wise Type I error rate.

Original significance level:

α = 0.05

Adjusted significance level:

α_adjusted = 0.05 / 3 = 0.01667

```python
# Statistical Analysis — Cell 37
# Pairwise Mann–Whitney U tests with Bonferroni correction

flows = ["Income", "Cost", "Balance Sheet"]

pairwise_results = []

for i in range(len(flows)):
    for j in range(i + 1, len(flows)):

        flow_1 = flows[i]
        flow_2 = flows[j]

        group_1 = df_stats.loc[
            df_stats["financial_flow"] == flow_1,
            "amount"
        ].dropna()

        group_2 = df_stats.loc[
            df_stats["financial_flow"] == flow_2,
            "amount"
        ].dropna()

        u_statistic, p_value = stats.mannwhitneyu(
            group_1,
            group_2,
            alternative="two-sided"
        )

        pairwise_results.append({
            "comparison": f"{flow_1} vs {flow_2}",
            "U_statistic": u_statistic,
            "p_value": p_value
        })

pairwise_analysis = pd.DataFrame(pairwise_results)

# Bonferroni correction
number_of_comparisons = len(pairwise_analysis)

pairwise_analysis["adjusted_alpha"] = (
    0.05 / number_of_comparisons
)

pairwise_analysis["significant"] = (
    pairwise_analysis["p_value"]
    < pairwise_analysis["adjusted_alpha"]
)

print("PAIRWISE MANN–WHITNEY U TESTS")
print("=" * 80)

display(pairwise_analysis)

print(
    f"\nBonferroni-adjusted significance level: "
    f"{0.05 / number_of_comparisons:.5f}"
)

print("\nInterpretation:")

for _, row in pairwise_analysis.iterrows():

    if row["significant"]:
        result = "Statistically significant"
    else:
        result = "Not statistically significant"

    print(
        f'{row["comparison"]}: '
        f'p-value = {row["p_value"]:.6e} → {result}'
    )
```

# 17. Effect Size Analysis

Statistical significance indicates whether an observed difference is unlikely to be explained by random variation.

Effect size evaluates the magnitude of the observed difference.

Effect sizes will be calculated for the major statistical comparisons performed in this analysis.

The purpose is to distinguish:

- Statistical significance
- Practical significance

A statistically significant result does not necessarily represent a large or practically important effect.

```python
# Statistical Analysis — Cell 39
# Effect size for 2022 vs 2023 Welch's t-test

mean_2022 = amount_2022.mean()
mean_2023 = amount_2023.mean()

std_2022 = amount_2022.std()
std_2023 = amount_2023.std()

n_2022 = len(amount_2022)
n_2023 = len(amount_2023)

# Pooled standard deviation
pooled_std = np.sqrt(
    (
        ((n_2022 - 1) * std_2022**2) +
        ((n_2023 - 1) * std_2023**2)
    )
    /
    (n_2022 + n_2023 - 2)
)

cohens_d = (
    (mean_2023 - mean_2022)
    / pooled_std
)

print("EFFECT SIZE — 2022 VS 2023")
print("=" * 70)

print(f"Mean difference: {mean_2023 - mean_2022:.6f}")
print(f"Pooled standard deviation: {pooled_std:.6f}")
print(f"Cohen's d: {cohens_d:.6f}")

print("\nInterpretation:")

abs_d = abs(cohens_d)

if abs_d < 0.2:
    interpretation = "Negligible effect"
elif abs_d < 0.5:
    interpretation = "Small effect"
elif abs_d < 0.8:
    interpretation = "Medium effect"
else:
    interpretation = "Large effect"

print(interpretation)
```

# 18. Financial Flow Pairwise Effect Size

The statistical significance tests identified significant differences between all three financial-flow pairs.

Effect size will now be calculated using rank-biserial correlation to assess the practical magnitude of these pairwise differences.

The effect size will be interpreted alongside the previously calculated p-values.

This distinction is important because a statistically significant result does not necessarily imply a large practical difference.

```python
# Statistical Analysis — Cell 41
# Rank-biserial effect size for pairwise Mann–Whitney tests

pairwise_effect_results = []

for i in range(len(flows)):
    for j in range(i + 1, len(flows)):

        flow_1 = flows[i]
        flow_2 = flows[j]

        group_1 = df_stats.loc[
            df_stats["financial_flow"] == flow_1,
            "amount"
        ].dropna()

        group_2 = df_stats.loc[
            df_stats["financial_flow"] == flow_2,
            "amount"
        ].dropna()

        u_statistic, p_value = stats.mannwhitneyu(
            group_1,
            group_2,
            alternative="two-sided"
        )

        n1 = len(group_1)
        n2 = len(group_2)

        # Rank-biserial correlation
        rank_biserial = (
            (2 * u_statistic) / (n1 * n2)
        ) - 1

        pairwise_effect_results.append({
            "comparison": f"{flow_1} vs {flow_2}",
            "U_statistic": u_statistic,
            "p_value": p_value,
            "rank_biserial_r": rank_biserial
        })

financial_flow_effect_sizes = pd.DataFrame(
    pairwise_effect_results
)

print("FINANCIAL FLOW PAIRWISE EFFECT SIZES")
print("=" * 80)

display(financial_flow_effect_sizes)

print("\nInterpretation guideline:")
print("|r| < 0.10  → Negligible")
print("|r| < 0.30  → Small")
print("|r| < 0.50  → Medium")
print("|r| >= 0.50 → Large")
```
