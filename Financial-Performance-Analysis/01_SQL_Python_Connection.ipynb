```python
# CONNECTION 
import pandas as pd
import mysql.connector

print("Pandas:", pd.__version__)
print("MySQL Connector: Successfully imported")
```

```python
# Step 2: Connect Python to MySQL

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_MYSQL_PASSWORD",
    database="financial_performance"
)

print("MySQL connection successful!")
```

```python
# Step 3: Verify project database tables

cursor = conn.cursor()

cursor.execute("SHOW TABLES")

for table in cursor:
    print(table[0])
```

```python
# Step 4: Check row counts for project tables

tables = [
    "budget",
    "customers",
    "financial_transactions",
    "financial_transactions_staging",
    "headcount",
    "vendors"
]

for table in tables:
    cursor.execute(f"SELECT COUNT(*) FROM {table}")
    count = cursor.fetchone()[0]
    print(f"{table}: {count} rows")
```

```python
# Step 5: Load financial transactions into Pandas

query = """
SELECT *
FROM financial_transactions
"""

df_transactions = pd.read_sql(query, conn)

print("Rows:", df_transactions.shape[0])
print("Columns:", df_transactions.shape[1])
```

```python
# Step 6: Inspect financial transactions

df_transactions.info()
```

```python
# Step 7: Preview loaded transaction data

df_transactions.head()
```

```python
# Step 8: Check Python data types

print(df_transactions.dtypes)
```

```python
# Step 9: Final SQL-Python connection validation

cursor.execute("""
    SELECT COUNT(*)
    FROM financial_transactions
""")

transaction_count = cursor.fetchone()[0]

print("Database:", "financial_performance")
print("financial_transactions rows:", transaction_count)

if transaction_count == len(df_transactions):
    print("SQL-Python validation: PASSED")
else:
    print("SQL-Python validation: FAILED")
```
